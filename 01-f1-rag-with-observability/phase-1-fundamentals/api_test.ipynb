Ergast Fetcher - Test

```python
from fetchers.ergast_fetcher import JolpicaFetcher

jolpica = JolpicaFetcher()
```

```python
# Cleanup
jolpica.close()
```

```python
# Driver standings after Monaco
driver_standings = jolpica.get_driver_standings(year=2024, round_number=8)
print('Driver standings after Monaco:\n')
for s in driver_standings[:10]:
    print(f'P{s.position}  {s.driver.code}  ({s.constructor.name:<20})  {s.points} pts')
```

```python
# Constructor standings after Monaco (round 8)
standings = jolpica.get_constructor_standings(year=2024, round_number=8)
print('Constructor standings after Monaco:\n')
for s in standings:
    print(f'P{s.position}  {s.constructor.name:<20}  {s.points} pts  ({s.wins} wins)')
```

```python
# Race results — Monaco 2024 is round 8
race = jolpica.get_race_results(year=2024, round_number=8)
print(f'Race: {race.raceName} | {race.date}')
print(f'Total classified: {len(race.results)} drivers\n')

# Top 5 finishers
for r in race.results[:5]:
    print(f'P{r.position}  {r.driver.code}  ({r.constructor.name})  —  {r.status}  —  {r.points} pts')
```
